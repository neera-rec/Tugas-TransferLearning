```python
from ultralytics import YOLO

# Memuat model dasar YOLOv8 versi 'nano' (paling ringan, cocok untuk sistem robotika)
model = YOLO('yolov8n.pt')

# Memulai proses training menggunakan dataset kamu
# Kita coba 50 putaran (epochs) dulu sebagai permulaan
results = model.train(data='/content/yolo_dataset/data.yaml', epochs=50, imgsz=640)
```
